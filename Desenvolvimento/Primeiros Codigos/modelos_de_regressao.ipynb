```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression # Modelo de regressão linea simples
from sklearn.model_selection import train_test_split # Ferramenta para separar nossos dados
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
from sklearn.metrics import root_mean_squared_error
```

```python
colunas = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 
           'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT', 'MEDV']

data = pd.read_csv('C:/Users/raphael.eugenio/Desktop/Raphael/Outros/train.csv', sep=',', index_col=0)
data.columns = colunas
```

```python
data.head()
```

```python
x = data.drop(["MEDV"], axis = 1) #separando os inputs de nosso target (MEDV->preco da casa)
y = data["MEDV"]

# Seperando em base de treino e teste com 30% dos dados reservados para avaliar
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=0)

# Definindo o algoritmmo utilizado
LinearRegressionModel = LinearRegression()

# Definindo o algoritmo com a base de treino
LinearRegressionModel.fit(x_train, y_train)

# Criando a predição que será usada na avalização
y_pred = LinearRegressionModel.predict(x_test)
```

```python
# Primeira avaliação R²
r2 = r2_score(y_test, y_pred)
print(r2)
```

```python
# Agora o R² Ajustado
def adjusted_r2(y_test, y_pred, x_train):
    
    adj_r2  = (1 - ((1 - r2_score(y_test, y_pred)) * (len(y_test) - 1)) /
               (len(y_test) - x_train.shape[1] - 1))
    
    return adj_r2
```

```python
adjusted_r2(y_test, y_pred, x_train)
```

```python
# Mean Squared Error or Erro Quadrático Médio (MSE)
mse = mean_squared_error(y_test, y_pred)
mse
```

```python
# Raiz do Erro Quadrático Médio (RMSE)
rmse = root_mean_squared_error(y_test, y_pred)
rmse
```
