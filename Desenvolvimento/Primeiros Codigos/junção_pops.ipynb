```python
import os
import re

from docx import Document
from docx.enum.text import WD_BREAK
from docxcompose.composer import Composer
```

```python
PASTA_ORIGEM = r"G:\Drives compartilhados\02 - DOCUMENTOS\Procedimentos operacionais (POPs)"

PASTA_SAIDA = os.path.join(PASTA_ORIGEM, "POPs_Consolidados")

os.makedirs(PASTA_SAIDA, exist_ok=True)
```

```python
arquivos = []

for arquivo in os.listdir(PASTA_ORIGEM):

    if not arquivo.lower().endswith(".docx"):
        continue

    m = re.search(r"POP\s*(\d+)", arquivo, re.IGNORECASE)

    if m:
        numero = int(m.group(1))
        arquivos.append((numero, arquivo))

arquivos.sort(key=lambda x: x[0])
```

```python
def grupo(numero):

    if numero <= 16:
        return "POP_01_A_16.docx"

    elif numero <= 32:
        return "POP_17_A_32.docx"

    elif numero <= 48:
        return "POP_33_A_48.docx"

    elif numero <= 64:
        return "POP_49_A_64.docx"

    elif numero <= 80:
        return "POP_65_A_80.docx"

    else:
        return "POP_81_A_98.docx"


grupos = {}

for numero, arquivo in arquivos:

    grupos.setdefault(grupo(numero), []).append(
        os.path.join(PASTA_ORIGEM, arquivo)
    )
```

```python
for nome_saida, lista_docs in grupos.items():

    print(f"\nCriando {nome_saida}")

    # Primeiro documento do grupo
    master = Document(lista_docs[0])

    composer = Composer(master)

    # Restantes
    for caminho in lista_docs[1:]:

        separador = Document()

        p = separador.add_paragraph()
        p.add_run().add_break(WD_BREAK.PAGE)

        titulo = separador.add_heading(
            os.path.basename(caminho),
            level=1
        )

        composer.append(separador)

        doc = Document(caminho)

        composer.append(doc)

    arquivo_saida = os.path.join(
        PASTA_SAIDA,
        nome_saida
    )

    composer.save(arquivo_saida)

    print(f"OK -> {arquivo_saida}")

print("\nFINALIZADO!")
```
