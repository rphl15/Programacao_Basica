```python
from pathlib import Path
from docx import Document
from docx.enum.text import WD_BREAK
from docxcompose.composer import Composer
```

```python
######### APENAS ALTERAR O CAMINHO DOS ARQUIVOS #############
PASTA = Path(
    r"G:\Drives compartilhados\02 - DOCUMENTOS\Políticas CMA"
)

arquivos = sorted(PASTA.glob("*.docx"))
```

```python
if not arquivos:
    raise Exception("Nenhum DOCX encontrado.")

master = Document(str(arquivos[0]))
composer = Composer(master)
```

```python
for arquivo in arquivos[1:]:

    separador = Document()

    p = separador.add_paragraph()
    p.add_run().add_break(WD_BREAK.PAGE)

    separador.add_heading(
        arquivo.stem,
        level=1
    )

    composer.append(separador)
    composer.append(Document(str(arquivo)))

saida = PASTA / "PROTOCOLOS_GERENCIADOS_CMA.docx"

composer.save(str(saida))

print(f"Arquivo criado: {saida}")
```
