```python
# Importando as Bibliotecas
import soccerdata as sd
import pandas as pd
from pathlib import Path

from soccerdata import ClubElo # As forças relativas da equipe em classificações Elo, na maioria das ligas europeias. Recalculado após cada rodada, inclui histórico.
from soccerdata import ESPN # Resultados históricos, estatísticas e escalações.
from soccerdata import FBref # Resultados históricos, escalações e estatísticas agregadas detalhadas para times e jogadores individuais com base em dados da Opta.
from soccerdata import MatchHistory # Resultados históricos, odds de apostas e estatísticas de partidas. O nível de detalhe depende da liga.
from soccerdata import Sofascore # Resultados, calendários, escalações e estatísticas detalhadas para times e jogadores individuais.
from soccerdata import SoFIFA # Pontuações detalhadas sobre todas as habilidades dos jogadores do EA Sports FC.
from soccerdata import Understat # Estatísticas avançadas como xG, xGBuildup e xGChain, e eventos de arremesso com valores xG associados para as principais ligas europeias.
from soccerdata import WhoScored # Resultados históricos, dados de prévia de partidas e dados detalhados de transmissões de eventos Opta para as grandes ligas.
```
